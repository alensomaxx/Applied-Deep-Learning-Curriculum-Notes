```python
import nltk
from nltk.corpus import treebank
from nltk.tag import hmm
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

# Download resources if not already available
nltk.download('treebank')
nltk.download('universal_tagset')

# Load dataset (Penn Treebank with universal POS tags)
train_data = treebank.tagged_sents(tagset='universal')[:3000]
test_data = treebank.tagged_sents(tagset='universal')[3000:]

# Train a Hidden Markov Model POS tagger
trainer = hmm.HiddenMarkovModelTrainer()
hmm_tagger = trainer.train_supervised(train_data)

# ---- Evaluate Accuracy ----
accuracy = hmm_tagger.evaluate(test_data)
print("Model Accuracy on Test Data: {:.2f}%".format(accuracy * 100))

# ---- Build Confusion Matrix ----
# Flatten the gold (true) and predicted tags
y_true = []
y_pred = []

for sent in test_data:
    words, true_tags = zip(*sent)
    predicted_tags = [tag for _, tag in hmm_tagger.tag(words)]
    y_true.extend(true_tags)
    y_pred.extend(predicted_tags)

# Create confusion matrix
labels = sorted(list(set(y_true)))  # unique POS tags
cm = confusion_matrix(y_true, y_pred, labels=labels)

# Plot confusion matrix
plt.figure(figsize=(10, 8))
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=labels)
disp.plot(cmap=plt.cm.Blues, xticks_rotation=45)
plt.title("Confusion Matrix for HMM POS Tagger")
plt.show()
```
