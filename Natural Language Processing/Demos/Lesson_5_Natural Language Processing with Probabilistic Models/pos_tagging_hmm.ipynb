```python
# POS Tagging with Hidden Markov Model (simplified example)
import nltk
from nltk.corpus import treebank

# Download if not already
nltk.download('treebank')
nltk.download('universal_tagset')

# Load corpus
train_data = treebank.tagged_sents(tagset='universal')[:3000]
test_data = treebank.tagged_sents(tagset='universal')[3000:]

# Train HMM Tagger
from nltk.tag import hmm
trainer = hmm.HiddenMarkovModelTrainer()
hmm_tagger = trainer.train_supervised(train_data)

# Test on a sentence
sentence = ["Time", "flies", "like", "an", "arrow"]
print(hmm_tagger.tag(sentence))
```
