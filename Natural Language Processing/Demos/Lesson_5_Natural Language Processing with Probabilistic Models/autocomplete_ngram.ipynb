```python
# Autocomplete with Bigram Language Model
from collections import defaultdict
import nltk

nltk.download('reuters')
from nltk.corpus import reuters

# Build corpus
words = [w.lower() for w in reuters.words() if w.isalpha()]

# Build bigram counts
bigram_counts = defaultdict(lambda: defaultdict(int))
for w1, w2 in nltk.bigrams(words):
    bigram_counts[w1][w2] += 1

def predict_next(word, top_n=3):
    candidates = bigram_counts[word]
    sorted_candidates = sorted(candidates.items(), key=lambda x: -x[1])
    return [w for w, _ in sorted_candidates[:top_n]]

print("Next word predictions for 'interest':", predict_next("interest"))
```
